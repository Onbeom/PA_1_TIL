1. 회전 행렬을 10만 번 누적해서 곱한 뒤, 직교성이 얼마나 무너졌는지를 수치로 측정하세요. 행렬과 그 전치의 곱이 단위행렬에서 얼마나 벗어났는지를 지표로 쓰면 됩니다.  
행렬 곱이 누적될수록 부동소수점 연산 오차가 커지며 직교성이 무너지는 것을 수치화하여 확인하고 행렬과 그 전치의 곱이 단위 행렬에서 벗어난 정도를 표시

```python
import sys
import os

sys.path.append(os.path.abspath(os.path.join(os.getcwd(),"..")))

import numpy as np
from src.rotation import rot_z, gram_schmidt
from src.vectors import dot_product, custom_cross

rng = np.random.default_rng(42)
angles =rng.uniform(-0.001, 0.001, 100000)

R_accum = np.eye(3)
for angle in angles:
    R_accum = R_accum @ rot_z(angle)

I_3 = np.eye(3)
error_before = np.max(np.abs((R_accum.T @R_accum) - I_3))
print(f"10만 번 누적 후 직교성 오차 (복구 전):{error_before}")
```

```python
print("[기록] 복구 전 오차가 존재함을 확인:", error_before > 0.0)
```

2. Gram-Schmidt 직교화를 직접 구현한 gram_schmidt 로 무너진 행렬을 복구하고, 복구 후 다시 지표를 측정해 전후를 비교하세요.
복구된 행렬이 진짜 회전행렬인지 확인하세요 — 직교성과 행렬식이 1임을 모두 만족해야 합니다.  
gram_schmidt 함수를 사용하여 오차가 누적되어 무너진 행렬 R_accum을 정규직교화 상태로 복구, 그 후 동일한 지표로 오차를 재측정하고 복구 전후의 오차 비율을 비교

```python
R_restored = gram_schmidt(R_accum)

error_after = np.max(np.abs((R_restored.T @ R_restored) - I_3))

improvement_ratio = error_before / error_after if error_after > 0 else float('inf')

print(f"재직교화 후 직교성 오차 (복구 후): {error_after}")
print(f"직교성 개선 배율: {improvement_ratio:.2e} 배 개선")
```

```python
is_restored = np.allclose(R_restored.T @ R_restored, I_3)
print("[검증] 재직교화 결과가 단위 행렬과 일치하는가:", is_restored)
assert is_restored
```
